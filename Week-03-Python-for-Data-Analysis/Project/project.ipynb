```python
import pandas as pd;
```

```python
data = pd.read_csv("data.csv")
```

```python
data.info();
```

```python
## Task 1 ; Handle Missing values

data.isnull().sum() ## date = 1 ; Calories = 2

data['Date'] = data['Date'].fillna('2020/12/22')

data['Calories'] = data['Calories'].fillna(data['Calories'].mean())
```

```python
## Task 2 ; Fitler row

data[data['Calories']>400]

data[data['Maxpulse']>120]

data[(data['Pulse']>100) & (data['Maxpulse']<140)]
```

```python
## Task 3 ; Add new column

data.loc[data["Maxpulse"] < 140, "Intensity"] = "Low"

data.loc[(data["Maxpulse"] >= 140) & (data["Maxpulse"] <= 160), "Intensity"] = "Medium"

data.loc[data["Maxpulse"] > 160, "Intensity"] = "High"
```
